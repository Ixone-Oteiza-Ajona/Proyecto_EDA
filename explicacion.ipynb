# EXPLICACIÓN DEL TEMA E HIPÓTESIS

El objetivo social y financiero consiste en incrementar el capital de las personas en el muy corto plazo (ventanas móviles de 7 a 30 días) con el propósito de generar riqueza y mitigar la vulnerabilidad económica.

# Planteamiento e hiótesis

Planteamiento de Hipótesis (Pre-Análisis)

Hipótesis Inicial ($H_0$ - Creencia popular): Para obtener grandes beneficios a muy corto plazo e incrementar rápidamente el capital de las personas, se debe invertir el 100 % del capital en los activos con mayor rendimiento nominal histórico reciente (criptoactivos como BTC-USD o renta variable tecnológica agresiva como QQQ).

Hipótesis Alternativa ($H_1$ - Descubrimiento empírico esperado): Invertir exclusivamente en activos de alta rentabilidad nominal a muy corto plazo expone a las personas a un Riesgo de Cola (Valor en Riesgo, VaR) y a caídas máximas (Maximum Drawdown) inasumibles para economías vulnerables. La mayor eficiencia en el crecimiento del capital a corto plazo (medida por el Ratio de Sharpe y el Ratio Sortino) se logra combinando un núcleo de preservación de capital (Renta Fija a corto plazo / Oro) con una exposición controlada al impulso (momentum) de renta variable.
